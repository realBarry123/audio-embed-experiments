```python
%matplotlib inline
import torch
from audembed import data
import matplotlib.pyplot as plt
try: weight = torch.load("experiments/readingframe/results/conv1_weight.pt")
except FileNotFoundError: weight = torch.load("results/conv1_weight.pt")
```

```python
data.plot_heatmap_2d(weight[:, 0, :])
```

```python
START = 0
N_PLOTS = 8
CHANNEL = 1
plt.rcParams['figure.figsize'] = [10, 16]
fig, axs = plt.subplots(N_PLOTS)
for i in range(N_PLOTS):
    axs[i].plot(weight[START+i, CHANNEL, :])
    axs[i].set_title(f"weight[{START+i}][{CHANNEL}]")
    axs[i].set_xlabel("kernel")
    axs[i].set_ylabel("weight")
    axs[i].set_ylim(-0.5, 0.5)
fig.tight_layout()
```

```python
START = 0
N_PLOTS = 8
plt.rcParams['figure.figsize'] = [10, 16]
fig, axs = plt.subplots(N_PLOTS)
weight_summed = weight.sum(dim=1)
for i in range(N_PLOTS):
    axs[i].plot(weight_summed[START+i, :])
    axs[i].set_title(f"weight[{START+i}][]")
    axs[i].set_xlabel("kernel")
    axs[i].set_ylabel("weight")
fig.tight_layout()
```
